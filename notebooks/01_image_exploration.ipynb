# Euclid image exploration / Exploración de imágenes Euclid
This starter notebook demonstrates image inspection using a public example image. Always record the source and licence.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

## Load a local image
Place an image you are permitted to use in `assets/images/` and update the path below.

```python
image_path = Path('../assets/images/example.jpg')
if image_path.exists():
    image = np.asarray(Image.open(image_path).convert('RGB'))
    print('Shape:', image.shape, 'dtype:', image.dtype)
    plt.figure(figsize=(12, 7))
    plt.imshow(image)
    plt.axis('off');
else:
    print('No local image yet. See assets/image_sources.csv and README for official archive links.')
```

## Questions to investigate
- What is the field of view and pixel scale?
- Which instrument/filter produced the image?
- Is this a colour composite? If so, how were channels mapped?
- What calibration, mask and background subtraction steps were applied?

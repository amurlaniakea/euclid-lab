# FITS image analysis / Análisis de imágenes FITS
A minimal workflow for a FITS image that you have downloaded from an archive and are allowed to analyse.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
from astropy.visualization import ImageNormalize, PercentileInterval, AsinhStretch
```

```python
fits_path = Path('../assets/images/example.fits')
if fits_path.exists():
    with fits.open(fits_path, memmap=False) as hdul:
        hdul.info()
        data = np.asarray(hdul[0].data, dtype=float)
        header = hdul[0].header.copy()
    print('Shape:', data.shape)
    print('BUNIT:', header.get('BUNIT', 'not specified'))
    if data.ndim == 2:
        norm = ImageNormalize(data, interval=PercentileInterval(99.5), stretch=AsinhStretch())
        plt.figure(figsize=(10, 8)); plt.imshow(data, origin='lower', cmap='gray', norm=norm)
        plt.colorbar(label=header.get('BUNIT', 'pixel value')); plt.show()
else:
    print('Add a suitable FITS image and update this path.')
```

## Scientific cautions
Do not interpret raw pixel values as flux without checking calibration and units. Inspect WCS, masks, exposure maps, variance, background treatment and instrument metadata. Preserve the original file and cite its archive identifier.
